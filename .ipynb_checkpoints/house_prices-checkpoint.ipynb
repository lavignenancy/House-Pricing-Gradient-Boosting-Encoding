## House Prices: Understanding & Comparing Gradient Boosting & Encoding Techniques

## Step 1: Setup
In this section, we import the necessary libraries and define the file paths for our datasets. Setting these variables upfront ensures clean, reusable code and eliminates the need to manually retype paths later.

```python
try:
    import xgboost
except ImportError:
    import sys
    !{sys.executable} -m pip install xgboost

import numpy as np
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import GradientBoostingRegressor, RandomForestRegressor
from sklearn.impute import SimpleImputer
from sklearn.metrics import root_mean_squared_error
from sklearn.model_selection import KFold, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, TargetEncoder
from xgboost import XGBRegressor

pd.set_option('display.max_columns', 100)

TRAIN_PATH = 'train.csv'
TEST_PATH = 'test.csv'
TARGET_COLUMN = 'SalePrice'
ID_COLUMN = 'Id'

RANDOM_STATE = 42
```

## Step 2: Loading the Data
Reading both training and test files into memory and checking their shapes to ensure all rows and features are loaded properly.

```python
train_df = pd.read_csv(TRAIN_PATH)
test_df = pd.read_csv(TEST_PATH)

print(f'train_df shape: {train_df.shape}')
print(f'test_df shape: {test_df.shape}')
train_df.head()
```

## Step 3: Target Variable Exploration & Transformation
House prices are naturally skewed to the right with large positive outliers. Taking the natural logarithm stabilizes variance and helps gradient boosting models optimize root mean squared error effectively.

```python
y = train_df[TARGET_COLUMN]
y_log = np.log1p(y)

print(f'Original SalePrice skew: {y.skew():.2f}')
print(f'Log-transformed SalePrice skew: {y_log.skew():.2f}')
```

## Step 4: Feature Separation & Column Identification
Separating the features from the target and identifying categorical versus numerical columns so each type receives the right preprocessing steps.

```python
X = train_df.drop(columns=[ID_COLUMN, TARGET_COLUMN])
X_test = test_df.drop(columns=[ID_COLUMN])

categorical_cols = X.select_dtypes(include=['object', 'category']).columns.tolist()
numerical_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()

print(f'Categorical features count: {len(categorical_cols)}')
print(f'Numerical features count: {len(numerical_cols)}')
```

## Step 5: Preprocessing Pipelines for Categorical Encoding
Building reusable preprocessing pipelines using ColumnTransformer. We compare three popular categorical encoding strategies alongside numerical imputation:

1. **Ordinal Encoding**: Converts text categories into integer ranks.
2. **One-Hot Encoding**: Expands each unique category into its own binary flag column.
3. **Target Encoding**: Replaces each category with the mean of the target variable for that level.

```python
num_transformer = SimpleImputer(strategy='median')


ordinal_cat_transformer = Pipeline(
    steps=[
        ('imputer', SimpleImputer(strategy='most_frequent')),
        (
            'encoder',
            OrdinalEncoder(
                handle_unknown='use_encoded_value', unknown_value=-1
            ),
        ),
    ]
)

preprocessor_ordinal = ColumnTransformer(
    transformers=[
        ('num', num_transformer, numerical_cols),
        ('cat', ordinal_cat_transformer, categorical_cols),
    ]
)


ohe_cat_transformer = Pipeline(
    steps=[
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=False)),
    ]
)

preprocessor_ohe = ColumnTransformer(
    transformers=[
        ('num', num_transformer, numerical_cols),
        ('cat', ohe_cat_transformer, categorical_cols),
    ]
)


target_cat_transformer = Pipeline(
    steps=[
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('encoder', TargetEncoder(smooth='auto', cv=5)),
    ]
)

preprocessor_target = ColumnTransformer(
    transformers=[
        ('num', num_transformer, numerical_cols),
        ('cat', target_cat_transformer, categorical_cols),
    ]
)
```

## Step 6: Cross-Validation & Model Evaluation
Setting up a 5-fold cross-validation scheme to evaluate our gradient boosting models with each encoding strategy on Root Mean Squared Error (RMSE) in log space.

```python
cv = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)


def evaluate_pipeline(pipeline, X_data, y_data):
    scores = cross_val_score(
        pipeline,
        X_data,
        y_data,
        cv=cv,
        scoring='neg_root_mean_squared_error',
        n_jobs=-1,
    )
    return -scores.mean(), scores.std()
```

## Step 7: Comparing Gradient Boosting Models with Different Encodings
Testing Scikit-Learn's GradientBoostingRegressor and XGBRegressor against each encoding approach to see which combination yields the lowest prediction error.

```python
models = {
    'Gradient Boosting (Sklearn)': GradientBoostingRegressor(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=4,
        random_state=RANDOM_STATE,
    ),
    'XGBoost': XGBRegressor(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=4,
        subsample=0.85,
        colsample_bytree=0.85,
        random_state=RANDOM_STATE,
    ),
}

preprocessors = {
    'Ordinal': preprocessor_ordinal,
    'One-Hot': preprocessor_ohe,
    'Target': preprocessor_target,
}

results = []

for m_name, model in models.items():
    for p_name, prep in preprocessors.items():
        pipe = Pipeline(steps=[('prep', prep), ('model', model)])
        mean_rmse, std_rmse = evaluate_pipeline(pipe, X, y_log)
        results.append({
            'Model': m_name,
            'Encoding': p_name,
            'Mean RMSE': round(mean_rmse, 4),
            'Std RMSE': round(std_rmse, 4),
        })

results_df = pd.DataFrame(results).sort_values(by='Mean RMSE')
results_df
```

## Step 8: Feature Importance Inspection
Examining which features contribute the most to the winning gradient boosting model's split decisions, demonstrating interpretability.

```python
best_pipeline = Pipeline(
    steps=[
        ('prep', preprocessor_ohe),
        (
            'model',
            XGBRegressor(
                n_estimators=300,
                learning_rate=0.05,
                max_depth=4,
                subsample=0.85,
                colsample_bytree=0.85,
                random_state=RANDOM_STATE,
            ),
        ),
    ]
)

best_pipeline.fit(X, y_log)

feature_names = best_pipeline.named_steps[
    'prep'
].get_feature_names_out()
importances = best_pipeline.named_steps['model'].feature_importances_

importance_df = (
    pd.DataFrame({'Feature': feature_names, 'Importance': importances})
    .sort_values(by='Importance', ascending=False)
    .head(15)
)

importance_df
```

## Step 9: Final Predictions & Submission Generation
Fitting the best performing pipeline on the entire training dataset, generating price predictions on the test set, transforming them back to original currency units using np.expm1, and saving submission.csv.

```python
best_pipeline.fit(X, y_log)

test_preds_log = best_pipeline.predict(X_test)
test_preds = np.expm1(test_preds_log)

submission_df = pd.DataFrame({
    ID_COLUMN: test_df[ID_COLUMN],
    TARGET_COLUMN: test_preds,
})

submission_df.to_csv('submission.csv', index=False)

print(f'Submission shape: {submission_df.shape}')
submission_df.head()
```
